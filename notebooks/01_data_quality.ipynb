```python
# MAPPING [1,2] TO [0,1] FOR CREDIT RISK FEATURE
df['Credit Risk'] = df['Credit Risk']-1

# saving it as csv file
df.to_csv("../data/in-between/german-credit_risk_in-between_dataset.csv", index=False)

```

```python
# PRINTING THE NUMBER OF DUPLICATES IN THE DATASET

print("Duplicates:", df.duplicated().sum())
```

```python
# DROPPING THE DUPLICATES IN THE DATASET
if df.duplicated().sum() > 0:
    df = df.drop_duplicates(inplace=True)
    print("Duplicates removed.")
else:
    print("No duplicates found.")
```
